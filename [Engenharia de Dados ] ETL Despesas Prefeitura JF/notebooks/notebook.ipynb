```python
import pandas as pd
from pathlib import Path
import numpy as np

pd.set_option('display.max_columns', None)
```

```python
path = Path("../data/despesas_26.csv")

df = pd.read_csv(path)
```

```python
df.info()
```

```python
df.head(1)
```

```python
colunas_drop= ['ug','funcao','subfuncao','programa', 'programa trabalho', 'categoria economica', 'grupo de natureza de despesa', 'modalidade de aplicacao', 'elemento de despesa orcamentaria','natureza de despesa com subelemento', 'fonte']

df.drop(columns=colunas_drop, inplace=True)
```

```python
df.head(1)
```

```python
df.info()
```

```python
numeric_cols = ['empenhado','liquidado','pago'] # mapeando variações comuns
actual_numeric_cols = [col for col in numeric_cols if col in df.columns]

for col in actual_numeric_cols:
    if df[col].dtype == 'str':
        # Remove pontos de milhar, substitui vírgula por ponto e converte para float
        df[col] = df[col].astype(str).str.replace('.', '', regex=False).str.replace(',', '.', regex=False)
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0.0)
```

```python
df.head(1)
```

```python
int_columns = ['cpf/cnpj', 'processo']

for col in int_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
        df[col] = df[col].astype('Int64')
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
linhas_drop=['unidade orcamentaria','nome modalidade de aplicacao']
df = df.dropna(subset=linhas_drop)
```

```python
df.head(1)
```

```python
colunas = ['data do empenho', 'data da liquidacao']
def transforma_para_data_correta(df):
    for col in colunas:
        if col in df.columns:
            # 1. Identifica e limpa os caracteres de ruído conhecidos
            # Passamos regex=False porque estamos buscando o texto exato
            df[col] = df[col].astype(str).str.strip().replace(['-', '#N/A', 'nan', 'NAN'], np.nan)
            
            # 2. Faz a conversão inteligente baseada no comportamento do dado
            df[col] = pd.to_datetime(
                df[col],
                format='%m/%d/%y',   # Analisa corretamente o padrão do arquivo
                errors='coerce'  # O que sobrou de texto inválido vira nulo com segurança
            )

            df[col] = df[col].dt.strftime('%d/%m/%Y').where(df[col].notna(), np.nan)
            
            total_nulos = df[col].isna().sum()
            print(f"Coluna '{col}' tratada. Total final de nulos reais: {total_nulos}")

# Lembre-se de recarregar o DataFrame original do CSV antes de rodar a função!
transforma_para_data_correta(df)
```
